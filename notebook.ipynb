```python
import chromadb
```

```python
client = chromadb.Client()
```

```python
collection = client.create_collection("my_collection")
```

```python
collection.add(
    documents=["This is a test document.", "This is another test document."],
    ids=["doc1", "doc2"]) # makes embedding / vectors and then adds to database
```

```python
collection.query(query_texts=["This is a test document."], n_results=2) # returns the most similar documents to the query
```
